```python
import pandas as pd

customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4, 5, 6],
    "total_spent": [350, 1200, 7500, 850, 3200, 9800]
})

customers
```

```python
def assign_segment(total_spent):
    if total_spent < 1000:
        return "Niska wartość"
    elif total_spent < 5000:
        return "Średnia wartość"
    else:
        return "Wysoka wartość"


customers["segment"] = customers["total_spent"].apply(assign_segment)

customers
```

```python
segment_summary = (
    customers.groupby("segment")
    .agg(
        liczba_klientow=("customer_id", "count"),
        suma_zakupow=("total_spent", "sum")
    )
)

segment_summary
```
